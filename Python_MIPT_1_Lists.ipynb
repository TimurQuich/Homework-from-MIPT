```python
from random import randint

N = 9
a, b = -30, 30
lst = [randint(a, b) for _ in range(N)]
print(f"Initial list: {lst}")
print(f"Minimal value: {min(lst)}")
print(f"Maximum value: {max(lst)}")
lst.sort()
print(f"Sorted list: {lst}")
print
if N % 2:
    med = lst[N//2]
else:
    med = (lst[(N-1)//2] + lst[N//2]) / 2
print(f"Median: {med}")
```

```python
from random import randint

N = 20
a, b = 0, 100 - 1
bin_range = 10
lst = [randint(a, b) for _ in range(N)]

# Bin Counts
bin_counts = [0 for _ in range(bin_range)]
for x in lst:
    bin_counts[x // 10] += 1

# Bin Probabilities
bin_prob = [bin_counts[i] / N for i in range(len(bin_counts))]

print(f"Initial list: {lst}\n")

for i in range(len(bin_counts)):
    print(f"Bin {i + 1} contains {bin_counts[i]} element(s). Probability: {bin_prob[i]}")
```

```python
from random import randint

N = int(input())
scalar = float(input())
a, b = -10, 10

vector_1 = [randint(a, b) for _ in range(N)]
vector_2 = [randint(a, b) for _ in range(N)]

sum_vector = [vector_1[i] + vector_2[i] for i in range(N)]
mult_vector = [vector_1[i] * vector_2[i] for i in range(N)]

norm_1 = sum([vector_1[i] ** 2 for i in range(N)]) ** 0.5
norm_2 = sum([vector_2[i] ** 2 for i in range(N)]) ** 0.5

if norm_1 >= norm_2:
    indx = 1
    vector_sc = [vector_1[i] * scalar for i in range(N)]
else:
    indx = 2
    vector_sc = [vector_2[i] * scalar for i in range(N)]

print(f"Vector 1: {vector_1}")
print(f"Vector 2: {vector_2}")

print(f"Component-wise sum: {sum_vector}")
print(f"Component-wise multiplication: {mult_vector}")

print(f"Multiplication of the vector {indx} by the scalar: {vector_sc}")
```

```python
from random import randint

M, N = 3, 5
V = 5

a, b = -20, 20

matrix = [[0] * N for _ in range(M)]
for i in range(M):
    for j in range(N):
        matrix[i][j] = randint(a, b)

vector = [[randint(a, b)] for _ in range(V)]

result = [[0] for _ in range(M)]
for i in range(M):
    c = 0
    for j in range(N):
        c += matrix[i][j] * vector[j][0]
    result[i][0] = c

print("Matrix:")
for row in matrix:
    print(*row)

print()

print("Vector")
for row in vector:
    print(*row)

print()

print("Multiplication of the matrix by the vector")
for row in result:
    print(*row)
```

```python
from random import randint

N = 10
a, b = -10, 10
lst = [randint(a, b) for _ in range(N)]
print(f"Initial list: {lst}")
for i in range(N):
    if lst[i] < 0:
        right_pos = None
        for j in range(i + 1, N):
            if lst[j] >= 0:
                right_pos = lst[j]
        left_pos = None
        for j in range(i - 1, -1, -1):
            if lst[j] >= 0:
                left_pos = lst[j]
        if right_pos and left_pos:
            lst[i] = (right_pos + left_pos) / 2

print(f"Processed list: {lst}")
```

```python
lst = [1, 2, 3, 4, 5, 6]
kernel = [0.5, 1, 0.5]

N = len(lst)
K = len(kernel)

R = N - K + 1
result = []

for i in range(R):
    current_sum = 0
    for j in range(K):
        current_sum += lst[i + j] * kernel[j]
    result.append(current_sum)

print(f"Initial list: {lst}")
print(f"Kernel: {kernel}")
print(f"Result: {result}")
```
