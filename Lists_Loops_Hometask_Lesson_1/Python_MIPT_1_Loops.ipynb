```python
# Task 2
count = iter(range(1, 12 + 1))
for i in range(5):
    for j in range(5):
        if (i + j) % 2:
            print(str(next(count)).ljust(3), end='')
        else:
            print('*'.ljust(3), end='')
    print()
```

```python
# Task 3
a, b = map(int, input().split())
while True:
    c = (a + b) // 2
    print("Is the number equal to {}?".format(c))
    answer = input()
    if answer == "Yes":
        result = c
        break
    else:
        print("Is the number less than {}?".format(c))
        answer = input()
        if answer == "Yes":
            b = c
        else:
            a = c
print("The answer is equal to {}".format(result))
```

```python
# Task 4
a = int(input())
m = a
while a > 0:
    a = int(input())
    m = a if a > m else m
print("The maximum number is equal to {}".format(m))
```
